```python
import sqlite3
import pandas as pd
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()
cursor.execute('''CREATE TABLE Rocket_Launches(
Launch_ID INTEGER PRIMARY KEY,
Rocket_Name TEXT,
Payload_Mass_KG INTEGER,
Launch_Site TEXT,
Status TEXT)''')
launches_data = [(1, 'Falcon 9', 5400, 'KSC LC 39A','Success'),
                 (2, 'Falcon Heavy', 12000, 'KSC LC 39A', 'Success'),
                 (3, 'Falcon 9', 2200, 'VAFB SLC 4E', 'Failure'),
                 (5, 'Falcon 9', 6000, 'CCAFS SLC 40', 'Success')]
cursor.executemany('INSERT INTO Rocket_Launches VALUES(?, ?, ?, ?, ?)', launches_data)
conn.commit()
print(f'SQL Database and Rocket_Launches table created successfully!')
```

```python
sql_query = 'SELECT * FROM Rocket_Launches;'
df_result = pd.read_sql_query(sql_query, conn)
print('---Active Database Records---')
print(df_result)
print('---Active Database Records---')
```

```python
filter_query = """SELECT Rocket_Name, Payload_Mass_KG, Status
FROM Rocket_launches
WHERE Status = 'Success'
ORDER BY Payload_Mass_KG DESC;"""
df_filtered = pd.read_sql_query(filter_query, conn)
print('---Heavest Successfull Rocket Launches---')
print(df_filtered)
```

```python
import pandas as pd
math_query = """SELECT
             COUNT(Launch_ID) AS Total_Launches,
             SUM(Payload_Mass_KG) AS Total_Payload_Mass,
             AVG(Payload_Mass_KG) AS Average_Payload_Mass,
             MAX(Payload_Mass_KG) AS Max_Payload_Mass,
             MIN(Payload_Mass_KG) AS Min_Payload_Mass
             FROM Rocket_Launches;"""
df_summary = pd.read_sql_query(math_query, conn)
print('---Summary Statistics---')
print(df_summary)
```

```python
import pandas as pd
group_query = """SELECT
             Launch_site,
             COUNT(Launch_ID) AS Number_of_flights,
             AVG(Payload_Mass_KG) AS Average_Site_payload
             FROM Rocket_Launches
             GROUP BY Launch_site;"""
df_grouped = pd.read_sql_query(group_query, conn)
print('---Launch Site Operaton Analytics---')
print(df_grouped)
```

```python
import pandas as pd
group_query = """SELECT
             Launch_site,
             COUNT(Launch_ID) AS Number_of_flights
             FROM Rocket_Launches
             WHERE Payload_Mass_KG > 5000
             GROUP BY Launch_site;"""
df_grouped = pd.read_sql_query(group_query, conn)
print('---Launch Site Operaton Analytics---')
print(df_grouped)
```

```python
import pandas as pd
group_query = """SELECT
             Status,
             COUNT(Launch_ID) AS Number_of_flights
             FROM Rocket_Launches
             WHERE Payload_Mass_KG > 5000
             GROUP BY Status;"""
df_grouped = pd.read_sql_query(group_query, conn)
print('---Status Operaton Analytics---')
print(df_grouped)

```

```python
import pandas as pd
group_query = """SELECT
             Status,
             COUNT(Launch_ID) AS Number_of_flights,
             AVG(Payload_Mass_KG) AS Average_Payload_Mass
             FROM Rocket_Launches
             WHERE Status = 'Success'
             GROUP BY Status;"""
df_grouped = pd.read_sql_query(group_query, conn)
print('---Status Operaton Analytics---')
print(df_grouped)
```

```python
cursor.execute('''
CREATE TABLE IF NOT EXISTS Site_Location(Site_code TEXT PRIMARY KEY,
                           State TEXT,
                           Country TEXT)''')
location_data = [('KSC LC 39A', 'California', 'United States'),
                 ('VAFB SLC 4E', 'Virginia', 'United States'),
                 ('CCAFS SLC 40', 'Texas', 'United States'),
                 ('CCAFS LC 40', 'California', 'United States')]
cursor.executemany('INSERT INTO Site_Location VALUES(?, ?, ?)', location_data)
conn.commit()
print("New 'Site_Location' table created and populated successfully!")
```

```python
import pandas as pd
join_query = """SELECT
             R.Rocket_Name,
             R.Payload_Mass_KG,
             R.Launch_Site,
             R.Status,
             S.State,
             S.Country
             FROM Rocket_Launches R
             JOIN Site_Location S
             ON R.Launch_Site = S.Site_code;"""

df_joined = pd.read_sql_query(join_query, conn)
print('---Unified Space & Location Dashboard---')
print(df_joined)
```
